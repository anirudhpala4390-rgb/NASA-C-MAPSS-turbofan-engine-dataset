Import Libraries


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

load datset

```python
df = pd.read_csv("/content/train_FD001.txt", sep=" ", header=None)
```

Nan's

```python
df.isna().sum()
```

```python
df.dropna(axis=1)
```

```python
df = df.dropna(axis=1)
```

Add column names

```python
columns = []
```

```python
columns.append("engine_id")
columns.append("cycle")
```

```python
for i in range(1, 4):
    name = "op_setting_" + str(i)
    columns.append(name)
```

```python
for i in range(1, 22):
    name = "sensor_" + str(i)
    columns.append(name)
```

```python
columns
```

```python
df
```

```python
df.columns
```

```python
df.head()
```

create Remaining Useful life (RUL)

```python
rul = df.groupby("engine_id")["cycle"].max().reset_index()
```

```python
rul.columns = ["engine_id", "max_cycle"]
```

```python
df = df.merge(rul, on="engine_id")
```

```python
df["RUL"] = df["max_cycle"] - df["cycle"]
```

Normalize the data

```python
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
```

```python
sensor_cols = []
```

```python
df.columns = columns
df.columns = df.columns.astype(str)
```

```python
for col in df.columns:
    if "sensor" in col:
        sensor_cols.append(col)
```

```python
df[sensor_cols] = scaler.fit_transform(df[sensor_cols])
```

```python
engine_1 = df[df["engine_id"] == 1]
```

Visualize the trend

```python
plt.plot(engine_1["cycle"], engine_1["sensor_2"])
plt.xlabel("Cycle")
plt.ylabel("Sensor 2")
plt.title("Sensor Trend Over Time")
plt.show()
```

Model building

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
```

```python
X = df[sensor_cols]
```

```python
X
```

```python
y = df["RUL"]
```

```python
y
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
```

```python
model = RandomForestRegressor()
```

```python
model.fit(X_train, y_train)
```

Evaluate

```python
from sklearn.metrics import mean_squared_error
```

```python
pred = model.predict(X_test)
```

```python
rmse = np.sqrt(mean_squared_error(y_test, pred))
```

```python
print("RMSE:", rmse)
```

Predictions

```python
print(pred[:10])
```

```python
results = pd.DataFrame({
    "Actual_RUL": y_test.values,
    "Predicted_RUL": pred
})
```

Visualization

```python
plt.scatter(y_test, pred)
plt.xlabel("Actual RUL")
plt.ylabel("Predicted RUL")
plt.title("Actual vs Predicted")
plt.show()
```
